```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 90, 90]
y = [10, 20, 9, -1]

# linestyle = -- - :
# marker = o * s
plt.plot(x, y, linestyle=':', marker='s')
plt.title("Panicking Rate")
plt.xlabel("month")
plt.ylabel("sale")

plt.show()
```
